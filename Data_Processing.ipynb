```python
def calculate_statistics(data, column):
    """Calculate summary statistics for a numeric column."""


    if data[column].dtype == 'object':
        values = data[column].str.replace('%', '', regex=False).astype(float).dropna()
    else :
        values = data[column].dropna()

    if values.empty:
        return {
            "Count": 0,
            "Average": None,
            "Highest": None,
            "Lowest": None
        }

    return {
        "Count": len(values),
        "Average": round(values.mean(), 2),
        "Highest": values.max(),
        "Lowest": values.min()
    }


statistics = calculate_statistics(df, "student_average")

print("STUDENT PERFORMANCE SUMMARY")
print("-" * 35)

for key, value in statistics.items():
    print(f"{key}: {value}")
```

```python
grade_counts = df["grade"].value_counts().sort_index()

grade_percentages = (
    df["grade"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
)

# Combine results into a summary table
grade_summary = pd.DataFrame({
    "Number_of_Students": grade_counts,
    "Percentage": grade_percentages
})

print("GRADE DISTRIBUTION")
display(grade_summary)
```
